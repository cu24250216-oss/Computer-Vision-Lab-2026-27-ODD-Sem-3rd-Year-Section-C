```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os

image_path = list(uploaded.keys())[0]

img = cv2.imread(image_path)

print("Image loaded successfully!")
```

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(10, 6))
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
height, width = img.shape[:2]
channels = img.shape[2]

print("Image Properties")
print("-------------------------")
print("Width:", width)
print("Height:", height)
print("Resolution:", width, "x", height)
print("Number of Channels:", channels)
print("Data Type:", img.dtype)
print("Total Pixels:", height * width)
```

```python
cv2.imwrite("image_output.jpg", img)
cv2.imwrite("image_output.png", img)

jpeg_size = os.path.getsize("image_output.jpg") / 1024
png_size = os.path.getsize("image_output.png") / 1024

print("JPEG saved successfully")
print("PNG saved successfully")
print(f"JPEG size: {jpeg_size:.2f} KB")
print(f"PNG size: {png_size:.2f} KB")
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

plt.figure(figsize=(10, 6))
plt.imshow(gray, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")
plt.show()
```

```python
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

hsv_display = cv2.cvtColor(hsv, cv2.COLOR_HSV2RGB)

plt.figure(figsize=(10, 6))
plt.imshow(hsv_display)
plt.title("HSV Image")
plt.axis("off")
plt.show()
```

```python
lab = cv2.cvtColor(img, cv2.COLOR_BGR2LAB)

lab_display = cv2.cvtColor(lab, cv2.COLOR_LAB2RGB)

plt.figure(figsize=(10, 6))
plt.imshow(lab_display)
plt.title("LAB Image")
plt.axis("off")
plt.show()
```

```python
resized = cv2.resize(img, None, fx=0.5, fy=0.5)

plt.figure(figsize=(8, 5))
plt.imshow(cv2.cvtColor(resized, cv2.COLOR_BGR2RGB))
plt.title("Resized Image")
plt.axis("off")
plt.show()
```

```python
height, width = img.shape[:2]

center = (width // 2, height // 2)

matrix = cv2.getRotationMatrix2D(center, 45, 1.0)

rotated = cv2.warpAffine(img, matrix, (width, height))

plt.figure(figsize=(10, 6))
plt.imshow(cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB))
plt.title("Rotated Image - 45 Degrees")
plt.axis("off")
plt.show()
```

```python
horizontal_flip = cv2.flip(img, 1)

plt.figure(figsize=(10, 6))
plt.imshow(cv2.cvtColor(horizontal_flip, cv2.COLOR_BGR2RGB))
plt.title("Horizontal Flip")
plt.axis("off")
plt.show()
```

```python
vertical_flip = cv2.flip(img, 0)

plt.figure(figsize=(10, 6))
plt.imshow(cv2.cvtColor(vertical_flip, cv2.COLOR_BGR2RGB))
plt.title("Vertical Flip")
plt.axis("off")
plt.show()
```

```python
negative = 255 - img

plt.figure(figsize=(10, 6))
plt.imshow(cv2.cvtColor(negative, cv2.COLOR_BGR2RGB))
plt.title("Negative Image")
plt.axis("off")
plt.show()
```

```python
height, width = img.shape[:2]

x1 = int(width * 0.25)
x2 = int(width * 0.75)
y1 = int(height * 0.25)
y2 = int(height * 0.75)

roi = img[y1:y2, x1:x2]

plt.figure(figsize=(10, 6))
plt.imshow(cv2.cvtColor(roi, cv2.COLOR_BGR2RGB))
plt.title("Region of Interest (ROI)")
plt.axis("off")
plt.show()
```

```python
plt.figure(figsize=(16, 10))

results = [
    (img_rgb, "Original"),
    (gray, "Grayscale"),
    (cv2.cvtColor(resized, cv2.COLOR_BGR2RGB), "Resized"),
    (cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB), "Rotated"),
    (cv2.cvtColor(horizontal_flip, cv2.COLOR_BGR2RGB), "Horizontal Flip"),
    (cv2.cvtColor(vertical_flip, cv2.COLOR_BGR2RGB), "Vertical Flip"),
    (cv2.cvtColor(negative, cv2.COLOR_BGR2RGB), "Negative"),
    (cv2.cvtColor(roi, cv2.COLOR_BGR2RGB), "ROI")
]

for i, (image, title) in enumerate(results):
    plt.subplot(2, 4, i + 1)

    if len(image.shape) == 2:
        plt.imshow(image, cmap="gray")
    else:
        plt.imshow(image)

    plt.title(title)
    plt.axis("off")

plt.tight_layout()
plt.show()
```
