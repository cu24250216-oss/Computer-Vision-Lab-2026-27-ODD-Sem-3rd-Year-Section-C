```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("Chest.png", cv2.IMREAD_GRAYSCALE)

plt.figure(figsize=(7, 5))
plt.imshow(image, cmap="gray")
plt.title("Original Chest X-ray")
plt.axis("off")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(image.ravel(), bins=256, range=[0, 256])

plt.title("Original Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.hist(image.ravel(), bins=256, range=[0, 256])
plt.title("Original Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")
plt.show()
```

```python
min_val = np.min(image)
max_val = np.max(image)

contrast_stretched = (
    (image.astype(np.float32) - min_val) /
    (max_val - min_val) * 255
).astype(np.uint8)

plt.figure(figsize=(7, 5))
plt.imshow(contrast_stretched, cmap="gray")
plt.title("Contrast Stretched Image")
plt.axis("off")
plt.show()
```

```python
equalized = cv2.equalizeHist(image)

plt.figure(figsize=(7, 5))
plt.imshow(equalized, cmap="gray")
plt.title("Histogram Equalized Image")
plt.axis("off")
plt.show()
```

```python
clahe = cv2.createCLAHE(
    clipLimit=2.0,
    tileGridSize=(8, 8)
)

clahe_image = clahe.apply(image)

plt.figure(figsize=(7, 5))
plt.imshow(clahe_image, cmap="gray")
plt.title("CLAHE Enhanced Image")
plt.axis("off")
plt.show()
```

```python
plt.figure(figsize=(16, 5))

plt.subplot(1, 4, 1)
plt.imshow(image, cmap="gray")
plt.title("Original")
plt.axis("off")

plt.subplot(1, 4, 2)
plt.imshow(contrast_stretched, cmap="gray")
plt.title("Contrast Stretching")
plt.axis("off")

plt.subplot(1, 4, 3)
plt.imshow(equalized, cmap="gray")
plt.title("Histogram Equalization")
plt.axis("off")

plt.subplot(1, 4, 4)
plt.imshow(clahe_image, cmap="gray")
plt.title("CLAHE")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.hist(image.ravel(), bins=256, range=[0, 256])
plt.title("Original Histogram")

plt.subplot(2, 2, 2)
plt.hist(contrast_stretched.ravel(), bins=256, range=[0, 256])
plt.title("Contrast Stretching Histogram")

plt.subplot(2, 2, 3)
plt.hist(equalized.ravel(), bins=256, range=[0, 256])
plt.title("Equalized Histogram")

plt.subplot(2, 2, 4)
plt.hist(clahe_image.ravel(), bins=256, range=[0, 256])
plt.title("CLAHE Histogram")

plt.tight_layout()
plt.show()
```
