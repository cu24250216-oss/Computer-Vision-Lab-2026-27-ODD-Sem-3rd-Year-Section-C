```python
image = cv2.imread("Screenshot 2026-08-14 193119.png", cv2.IMREAD_GRAYSCALE)

if image is None:
    print("Image not found!")
else:
    print("Image loaded successfully!")
    print("Image shape:", image.shape)
```

```python
plt.figure(figsize=(6, 5))

plt.imshow(image, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.show()
```

```python
average_filter = cv2.blur(image, (5, 5))

plt.figure(figsize=(6, 5))
plt.imshow(average_filter, cmap="gray")
plt.title("Average Filter")
plt.axis("off")
plt.show()
```

```python
gaussian_filter = cv2.GaussianBlur(image, (5, 5), 0)

plt.figure(figsize=(6, 5))
plt.imshow(gaussian_filter, cmap="gray")
plt.title("Gaussian Filter")
plt.axis("off")
plt.show()
```

```python
median_filter = cv2.medianBlur(image, 5)

plt.figure(figsize=(6, 5))
plt.imshow(median_filter, cmap="gray")
plt.title("Median Filter")
plt.axis("off")
plt.show()
```

```python
laplacian = cv2.Laplacian(image, cv2.CV_64F)
laplacian = cv2.convertScaleAbs(laplacian)

plt.figure(figsize=(6, 5))
plt.imshow(laplacian, cmap="gray")
plt.title("Laplacian Filter")
plt.axis("off")
plt.show()
```

```python
sobel_x = cv2.Sobel(image, cv2.CV_64F, 1, 0, ksize=3)
sobel_x = cv2.convertScaleAbs(sobel_x)

plt.figure(figsize=(6, 5))
plt.imshow(sobel_x, cmap="gray")
plt.title("Sobel X - Vertical Edges")
plt.axis("off")
plt.show()
```

```python
sobel_y = cv2.Sobel(image, cv2.CV_64F, 0, 1, ksize=3)
sobel_y = cv2.convertScaleAbs(sobel_y)

plt.figure(figsize=(6, 5))
plt.imshow(sobel_y, cmap="gray")
plt.title("Sobel Y - Horizontal Edges")
plt.axis("off")
plt.show()
```

```python
sobel_combined = cv2.addWeighted(
    sobel_x, 0.5,
    sobel_y, 0.5,
    0
)

plt.figure(figsize=(6, 5))
plt.imshow(sobel_combined, cmap="gray")
plt.title("Combined Sobel")
plt.axis("off")
plt.show()
```

```python
plt.figure(figsize=(14, 8))

plt.subplot(2, 4, 1)
plt.imshow(image, cmap="gray")
plt.title("Original")
plt.axis("off")

plt.subplot(2, 4, 2)
plt.imshow(average_filter, cmap="gray")
plt.title("Average")
plt.axis("off")

plt.subplot(2, 4, 3)
plt.imshow(gaussian_filter, cmap="gray")
plt.title("Gaussian")
plt.axis("off")

plt.subplot(2, 4, 4)
plt.imshow(median_filter, cmap="gray")
plt.title("Median")
plt.axis("off")

plt.subplot(2, 4, 5)
plt.imshow(laplacian, cmap="gray")
plt.title("Laplacian")
plt.axis("off")

plt.subplot(2, 4, 6)
plt.imshow(sobel_x, cmap="gray")
plt.title("Sobel X")
plt.axis("off")

plt.subplot(2, 4, 7)
plt.imshow(sobel_y, cmap="gray")
plt.title("Sobel Y")
plt.axis("off")

plt.subplot(2, 4, 8)
plt.imshow(sobel_combined, cmap="gray")
plt.title("Sobel Combined")
plt.axis("off")

plt.tight_layout()
plt.show()
```

**Questions**

**Q1. What is spatial filtering? How is it used in digital image processing?**

Answer:
Spatial filtering is an image processing technique in which the value of a pixel is modified using its neighboring pixels. A small matrix called a kernel or mask is moved over the image.

Spatial filtering is used for:

Noise removal
Image smoothing
Sharpening
Edge detection
Image enhancement

Example: Gaussian filter is used for smoothing, while Sobel filter is used for edge detection.

**Q2. Differentiate between Low-Pass Filters and High-Pass Filters with suitable examples.**
Answer:

Low-Pass Filter	High-Pass Filter
Removes high-frequency components	Enhances high-frequency components
Used for smoothing	Used for edge detection
Reduces noise	Highlights edges
Produces a smoother image	Enhances fine details
Examples: Average, Gaussian, Median	Examples: Sobel, Laplacian

Low-Pass → Smoothing and Noise Reduction

High-Pass → Edge Detection and Sharpening

**Q3. Compare Average Filter, Gaussian Filter, and Median Filter based on noise reduction and edge preservation.**
Average Filter
Answer:
It calculates the average value of neighboring pixels. It reduces noise but can blur edges.

Gaussian Filter

It gives more importance to pixels near the center and provides smooth noise reduction.

Median Filter

It replaces a pixel with the median value of its neighborhood. It is especially effective for salt-and-pepper noise and preserves edges well.

**Q4. Why is the Median Filter particularly effective for removing salt-and-pepper noise?**
Answer:
Salt-and-pepper noise consists of random black and white pixels.

The Median filter replaces a pixel with the median value of its neighboring pixels. Extreme black and white values have less effect on the median.

Therefore, the Median filter removes salt-and-pepper noise while preserving important edges.

**Q5. Explain the role of convolution in spatial filtering.**
Answer:
Convolution is a mathematical operation used in spatial filtering. A small matrix called a kernel is moved over the image.

The kernel values are multiplied with the corresponding image pixels and then added to produce the new pixel value.

**Q6. What is the purpose of Sobel and Laplacian operators in edge detection?**

Answer:
Sobel and Laplacian operators are High-Pass filters used for detecting edges.

Sobel Operator
Sobel X detects vertical edges.
Sobel Y detects horizontal edges.
Laplacian Operator

The Laplacian operator uses the second derivative to detect rapid changes in intensity and edges in multiple directions.

Therefore:

Sobel → Directional edge detection

Laplacian → Multi-directional edge detection

**Q7. Why are Gaussian filters considered smoothing filters in image processing?**

Answer:
Gaussian filters are smoothing filters because they reduce sudden changes in pixel intensity.

They assign higher weights to pixels near the center and lower weights to pixels farther away.

Gaussian filtering is used to:

Reduce noise
Remove small details
Smooth images
Prepare images for edge detection

**Q8. Benefits and limitations of frequency-domain filtering**

Answer:
Benefits:

Removes periodic noise effectively.
Separates low and high frequencies.
Useful for image enhancement.

Limitations:

More complex than spatial filtering.
Requires Fourier Transform.
Needs more computation.

**Q9. Two real-world applications of spatial filtering**

Answer:

Medical Imaging – Noise removal and image enhancement in X-rays, MRI, and CT scans.
Computer Vision – Edge detection in face detection, CCTV, and autonomous vehicles.

**Q10. Spatial-domain vs Frequency-domain filtering.**
Answer
Spatial Domain	Frequency Domain
Works directly on pixels	Works on frequency components
Uses kernels/masks	Uses Fourier Transform
Simple and easy	More complex
Used for smoothing and edge detection	Used for noise removal and enhancement

In short:
Spatial → Pixels + Kernel
Frequency → Fourier Transform
